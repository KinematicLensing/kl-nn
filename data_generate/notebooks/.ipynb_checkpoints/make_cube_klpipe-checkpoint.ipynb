```python
# common test parameters
#these are all the things I should need to make a datacube? no I still ought to specify emission lines or something

_IMAGE_PARS = ImagePars(shape=(64, 64), pixel_scale=0.05, indexing='ij') # pixel_scale=0.11

_VEL_PARS = {
    'cosi': 0.5,
    'theta_int': 0.7,
    'g1': 0.0,
    'g2': 0.0,
    'v0': 10.0,
    'vcirc': 200.0,
    'vel_rscale': 0.5,
}

_INT_PARS = {
    'cosi': 0.5,
    'theta_int': 0.7,
    'g1': 0.0,
    'g2': 0.0,
    'flux': 100.0,
    'int_rscale': 0.3,
    'int_h_over_r': 0.1,
    'int_x0': 0.0,
    'int_y0': 0.0,
}

_SHARED_PARS = {'cosi', 'theta_int', 'g1', 'g2'}
```

```python
#meta pars? to make intensity model
#meta pars takes a dictionary, something like this
```

```python
int_model = InclinedExponentialModel(meta_pars=_INT_PARS)
int_model
```

```python
vel_model = CenteredVelocityModel(meta_pars=_VEL_PARS)
vel_model
```

```python
theta_vel = vel_model.pars2theta(_VEL_PARS)
theta_int = int_model.pars2theta(_INT_PARS)
```

```python
def cube_pars(): #this thing needs image_pars and such
    z = 1.0
    lam_center = HALPHA.lambda_rest * (1 + z)
    dlam = 2.5
    return CubePars.from_range(_IMAGE_PARS, lam_center - dlam, lam_center + dlam, 0.08)
```

```python
cubepars=cube_pars()
```

```python
#ingredients for Spectral Model: SpectralConfig, IntensityModel, VelocityModel
#cubepars can be made from image pars and wavelength range

#need a specified intensity model
config = SpectralConfig(lines=(halpha_line(),))
sm = SpectralModel(config, int_model, vel_model) #I notice that this thing alone does not take cubepars
```

```python
#theta_spec?
theta_spec = jnp.array([1.0, 5., 1., 0.01])  # z, vel_disp, Ha_flux, Ha_cont
```

```python
datacube = sm.build_cube(theta_spec, theta_vel, theta_int, cubepars)
```
