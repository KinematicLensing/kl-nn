```python
import torch
import matplotlib.pyplot as plt
```

```python
unif = torch.rand(2, 10000)
```

```python
whr = torch.where(unif[0] > 0.5)[0]
weights = torch.full_like(unif, 1.)
weights[:, whr] = 0
weights /= weights.sum()
```

```python
resamp_idx = torch.multinomial(weights, 10000, replacement=True)
resamp = unif[:, resamp_idx]
```

```python
plt.figure(figsize=(6,6), dpi=300)
plt.scatter(unif[0], unif[1], s=0.1)
plt.xlim(0, 1)
plt.ylim(0, 1)
```

```python
plt.figure(figsize=(6,6), dpi=300)
plt.scatter(resamp[0], resamp[1], s=0.1)
plt.xlim(0, 1)
plt.ylim(0, 1)
```
